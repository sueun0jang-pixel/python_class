## if2.py
##### if ~ else 구문 (양자택일)에 대한 예제 프로그램입니다.
##### 성적이 0 ~ 100 사이의 정수가 아닌 경우는 '성적을 정확히 입력하세요'로 출력하고
##### 성적이 0 ~ 100 사이의 정수인 경우 60점 이상인 경우 합격, 그렇지 않는 경우 불합격을 출력하시오.

```python
score = float(input("성적을 입력하세요"))

try:
    if isinstance(score, int) or isinstance(score, float):
        if isinstance(score, int):
            if 80 <= score <= 100:
                print('합격입니다.')
            elif score < 80:
                print("불합격입니다.")
            elif score > 100 or score <0:
                print("성적은 0 ~ 100 사이의 정수여야 합니다.")
        elif isinstance(score, float) :
            print("성적을 정확히 입력하세요")
    else:
        print("숫자만 입력해주세요.")
except ValueError:
    print("성적을 정확히 입력하세요")
```

```python
score = int(input("성적을 입력하세요."))

if score < 80 and score >=60:
    print('합격입니다.')
elif score >=80:
    print("합격했습니다.", end=' ')
    print("장학금을 받을 수 있습니다.")    
elif score <60:
    print('불합격입니다.')

```
