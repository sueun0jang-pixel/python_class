## String3.py
##### sequence : 인덱싱과 슬라이싱에 대한 예제입니다.

```python
str = "Monty python"

print(str[0])
print(str[-1])
```

```python
str = "Monty python"

print(str[0:5])
print(str[:5])
print(str[6:])
```

```python
date_weather = '20250722Sunny'

#날짜(date)와 날씨(weather)를 출력하시오.
print(date_weather[:8])
print(date_weather[8:])

```
