## range1.py
##### 반복문에 대한 예제 프로그램 입니다.
##### range(start=0, stop, step=1) 함수에 사용

```python
for i in range(1, 11, 1):
    print(i, end='\t')

print()

for i in range(10):
    print(i, end='\t')
```

```python
user_num = int(input('단을 입력하세요.'))

print(f'구구단을 외자!\n{user_num}단 입니다.')
print()

for i in range(1, 10, 1):
    print(f'{user_num} * {i} = {user_num * i}')
```
