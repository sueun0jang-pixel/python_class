##### 정수를 입력 받아서 짝수인 경우 True, 홀수인 경우는 False를 반환하는 함수를 정의해보세요.

```python
def odd(x):
    if x%2 == 0:
        return 'even'
    else:
        return 'odd'

user_num = int(input("정수를 입력해주세요."))
result = odd(user_num)

if result == 'even':
    print(f'입력하신 {user_num}은 짝수입니다.')
elif result == 'odd':
    print(f'입력하신 {user_num}은 홀수입니다.')
```

```python
# is_even 함수 사용할 것
# 프로그램 종료 : q
# 짝수인지 확인할 숫자를 입력하세요 : 2
# 2는 짝수입니다.
# 짝수인지 확인할 숫자를 입력하세요 : 5
# 2는 짝수가 아닙니다.

is_even = True

def is_even(x):
    if x%2 == 0:
        return is_even
    else:
        is_even=False
        return is_even


while True:
    answer = input('짝수인지 확인할 숫자를 입력하세요.(종료:q)')

    if answer == 'q':
        break

    num = int(answer)

    if is_even(num):
        print(f'{num}은 짝수입니다.')
    else:
        print(f'{num}은 홀수입니다.')
```
