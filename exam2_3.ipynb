##### 물건을 구입할 때 구입액이 10만원 이상이면, 5%의 할인을 받을 수 있다.
##### 사용자에게 우선 구입금액을 물어보고 할인 금액과 최종 지불 금액을 출력하는 예제 프로그램을 만드시오.

```python
purchace = int(input("구입 금액을 입력하세요."))

if purchace >= 100000:
    discount = purchace * 0.005
elif purchace < 100000:
    discount = 0
    
final_price = purchace - discount

print(f'할인 금액은 {discount}원 입니다.')
print(f'최종 지불 금액은 {final_price}원 입니다.')
```

```python
class purchace:
    def __init__(self, amount):
        self.amount = amount
        print(self.amount)

    def calculate_discount(self):
        if self.amount >= 100000:
            discount = self.amount * 0.05
            final_price = self.amount - discount
            return f'최종 지불 금액은 {final_price}원 입니다.'
        else:
            return f'최종 지불 금액은 {self.amount}원 입니다.'
purchace_amount = int(input("구입 금액을 입력하세요."))

def main():
    user_purchace = purchace(purchace_amount)
    print(user_purchace.calculate_discount())
```
