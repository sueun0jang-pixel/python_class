## random1.py
##### 랜덤 함수에 대한 예제 프로그램입니다.

```python
import random

for i in range(100):
    num = random.randint(1, 100)
print(num)
```

```python
for i in range(100):
    ch = random.choice('abcdefghijklmnopqrstupwvxyz') # 시퀀스
print(ch)
```

```python
for i in range(100):
    fruit = random.choice(['apple', 'banana', 'kiwi', 'orange'])
print(fruit)
```
