## if1.py
##### if문에 대한 예제 프로그램입니다.

```python
score = 100

if score >=60 :
    print('합격입니다.')
else :
    print('end')

score = 50

if score >=60:
    print('합격입니다.')
print('end')
```

```python
score = int(input('성적을 입력하세요.'))

if score >= 60:
    print('합격입니다.')

print('end')
```

```python
score = int(input('성적을 입력하세요.'))

if score >= 90:
    print('합격입니다.')
    print('장학금을 받을 수 있습니다.')

print('end')
```
